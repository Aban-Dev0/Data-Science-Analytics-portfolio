# Exploratory Data Analysis : Student Performance Dataset
### I have this dataset (student_data) and I will clean , change and find correlations 

```python
import pandas as pd

# create a simple pandas dataframe (table)
student_data = {
    'name :': ["kofi", "ama", "kwame"],
    "score :": [85, 92 , 78]
}
my_table = pd.DataFrame(student_data)
print(my_table)
```

```python
# add a new column with a missing value (none) to stimulate real world messu data 
my_table['study_hours'] = [5 , None , 3]

print("table with missing data :")
print(my_table)


#command pandas to instantly find and remove any rows containing missing info
cleaned_table = my_table.dropna()
print("\ncleaned table :")
print(cleaned_table)
```

```python
# calculate the average study times using the existing numbers(ignores the blank spaces)
average_hours = my_table['study_hours'].mean()
print('average hours :' , average_hours)

#command pandas to the missing spot with the average instead of deleting the row 
filled_table = my_table.fillna(average_hours)
print('salvaged table :\n' , filled_table)
```

```python
#cslculate the correlation score (a number between -1 and 1.0)
#1.0 means the perfectly increase together , 0 means no connection

connections = filled_table.corr(numeric_only=True)
print('mathematical connections')
print(connections)
```
